```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = LabelEncoder()
```

```python
data["Label"] = label_encoder.fit_transform(data["variety"])
```

```python
data.head(15)
```

```python
data.head(60).tail()
```

```python
categories = list(label_encoder.inverse_transform([i for i in range(3)]))
```

```python
print(categories)
```

```python
from sklearn.model_selection import train_test_split
```

```python
data.head()
```

```python
data.drop(["variety"], axis=1, inplace=True)
```

```python
data.head()
```

```python
X,y = data.iloc[:,:-1], data.iloc[:, -1]
```

```python
data.head()
```

```python
y.head()
```

```python
X_tran, X_test, y_tran, y_test = train_test_split(X,y, test_size=0.2, random_state=0)
```

```python
y_tran.info()
```

```python
X_test.info()
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
classifier = DecisionTreeClassifier(max_depth=1, criterion="gini", random_state=0)
```

```python
classifier.fit(X_tran, y_tran)
```

```python
train_acc = classifier.score(X_tran, y_tran)
```

```python
print("Train Accuracy: ", train_acc)
```

```python
test_acc = classifier.score(X_test, y_test)
```

```python
print("Test Accuracy: ", test_acc)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test, classifier.predict(X_test)))
```

```python
from sklearn.metrics import confusion_matrix
```

```python
conf_mat = confusion_matrix(y_test, classifier.predict(X_test))
```

```python
conf_mat
```

```python
ax = sns.heatmap(conf_mat, annot=True, square=True, cmap="Blues", fmt="d")
ax.xaxis.set_ticklabels(categories)
ax.yaxis.set_ticklabels(categories)
ax.set_xlabel("Predicted")
ax.set_ylabel("Actual")
plt.show()
```
